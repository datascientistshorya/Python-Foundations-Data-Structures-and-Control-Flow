#### Q86. Create a list containing "High" for sales >= 10,000 and "Low" otherwise.

```python
sales=[8000,12000,10000,3000,20000]
status=[]
for i in sales:
    if i>=10000:
        status.append('High')
    else:
        status.append('Low')
status
```

```python
#using list comprehension
sales = [8000, 12000, 10000, 3000, 20000]
status = ["High" if i >= 10000 else "Low" for i in sales]
status
```

```python
customers = [
    {"name": "A", "spend": 50000},
    {"name": "B", "spend": 30000},
    {"name": "C", "spend": 80000}
]
names=[i['name'] for i in customers if i['spend']>=50000]
names
```

#### Q88. Create a list of squared values only for positive numbers.

```python
numbers=[2,-3,-4,5,-7,0]
squares =[n*n for n in numbers if n>0]
squares
```

```python
number=[12,2,34,6,78,5]
squares=[n*n for n in number if n>10]  
squares
```

#### Q90. Explain when a normal loop is preferable to a comprehension.
answer : A **normal loop is preferable** when the logic is complex, involves multiple steps or conditions, requires `break`/`continue`, or needs side effects, because it is generally easier to read and debug than a comprehension.

#### Q21. Write nested conditions that first check whether a customer is active and then check whether their spend exceeds ₹50,000.

```python
status=input('enter customer status....active/inactive?')
spend=int(input('enter customer spend'))
if status=='active':
    if spend>50000:
        print('Valuable Customer')
    else:
        print('worth improving')
        
else:
    print('worth improving')
```

```python
orders = 5
spend = 30000
if orders>=5:
    if spend>=30000:
        print('VIP')
    else:
        print('High Value')
else:
    print('Regular')
        
```

#### Q35. Calculate the average sales using a loop.

```python
sum=0
sales = [1000, 2500, 3000, 1500]
length=len(sales)
for i in sales:
    sum+=i
print('Average Sales ',sum/length)
```

```python
sales = [2000, 7000, 4500, 9000, 3000, 8000]
counter=0
for i in sales:
    if i<5000:
        counter+=1

print('Sales greater than 5000 ',counter)
```

```python
statuses = ["Completed", "Cancelled", "Completed", "Completed", "Cancelled"]

counter=0
for i in statuses:
    if i=='Completed':
        counter+=1

print('Total completed sales ',counter)
```

```python
orders = [
    ("Completed", 5000),
    ("Cancelled", 3000),
    ("Completed", 7000),
    ("Cancelled", 2000)
]
sum=0
for i in orders:
    if i[0]=='Completed':
        sum+=i[1]

print('Sales from completed orders ',sum)
    
```

```python
max_s=0
sales = [4500, 12000, 7000, 18000, 9000]
for i in sales:
    if i>max_s:
        max_s=i

print('Largest sale ',max_s)
```

#### Q96. Calculate the cancellation rate from the same dataset.

```python
transactions = [
    {"amount": 5000, "status": "Completed"},
    {"amount": 7000, "status": "Cancelled"},
    {"amount": 12000, "status": "Completed"},
    {"amount": 3000, "status": "Completed"}
]
can=0
for i in transactions:
    if i['status']=='Cancelled':
        can+=1
print('Cancellation Rate: ',can/len(transactions)*100,'%')
    
```

```python
funnel = {
    "view_item": 10000,
    "add_to_cart": 3000,
    "begin_checkout": 1500,
    "purchase": 900
}

stages = list(funnel.keys())

largest_drop = 0
largest_drop_stage = ""

for i in range(1, len(stages)):
    previous = funnel[stages[i - 1]]
    current = funnel[stages[i]]

    drop = (previous - current) / previous * 100

    if drop > largest_drop:
        largest_drop = drop
        largest_drop_stage = stages[i]

print("Largest drop:", largest_drop_stage)
print("Drop percentage:", largest_drop)
```

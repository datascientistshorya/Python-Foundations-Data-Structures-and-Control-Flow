#### Q68. Count how many values exceed 50.

```python
data = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]
count=0
for i in data:
    for j in range(3):
        if i[j]>50:
            count+=1

print(count)
        
```

#### Q69. Calculate the average of all values in the nested list.

```python
data = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]
total=0
for i in data:
    for j in range(3):
        total+=i[j]
print(total/9)
        
```

#### Q70. Explain why nested loops can become expensive with large datasets.
answer: Nested loops can become expensive because **for every iteration of the outer loop, the inner loop runs again**,
so the number of operations can grow very quickly—for example, `1000 × 1000 = 1,000,000` iterations.

```python
categories = [
    "Laptop",
    "Mobile",
    "Laptop",
    "Tablet",
    "Mobile",
    "Laptop"
]
frequency = {}
for category in categories:
    if category in frequency:
        frequency[category] += 1
    else:
        frequency[category] = 1

print(frequency)
```

#### Q77. Count completed and cancelled orders separately.

```python
orders = [
    {"status": "Completed", "amount": 5000},
    {"status": "Cancelled", "amount": 3000},
    {"status": "Completed", "amount": 7000}
]
com=0
can=0
for order in orders:
    if order['status']=='Completed':
        com+=1
    else:
        can+=1
print('Completed :',com,'Cancelled :',can)
```

#### Q78. Calculate average completed order value.



```python
orders = [
    {"status": "Completed", "amount": 5000},
    {"status": "Cancelled", "amount": 3000},
    {"status": "Completed", "amount": 7000}
]
total=0
count=0
for i in orders:
    if i['status']=='Completed':
        total+=i['amount']
        count+=1
print('Avg completed order: ',total/count)
    
```

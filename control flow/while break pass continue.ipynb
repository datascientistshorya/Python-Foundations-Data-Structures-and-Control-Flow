#### Q42. Write a while loop that prints numbers from 10 down to 1.

```python
i=10
while i>0:
    print(i)
    i-=1
```

#### Q43. Calculate the sum of numbers from 1 to 100 using a while loop.

```python
total=0
i=1
while i<101:
    total+=i
    i+=1
print(total)
```

# Topic 6 — break, continue, pass

```python
values = [10, 20, 30, -5, 40, 50]
for i in values:
    if i<0:
        break
    else:
        print(i)
```

```python
statuses = ["Completed", "Cancelled", "Completed", "Cancelled"]
for i in statuses:
    if i=='Cancelled':
        continue
    else:
        print(i)
```

```python
values = [10, 20, -30, -5, 40, 50]
for i in values:
    if i<0:
        continue
    else:
        print(i)
```

```python
customers = [
    ("A", 20000),
    ("B", 35000),
    ("C", 75000),
    ("D", 90000)
]

for customer, spending in customers:
    if spending > 50000:
        print(customer)
        break
    
```

```python
scores = [0.91, None, 0.85, None, 0.93]
for i in scores:
    if  i== None:
        continue
    else:
        print(i)
```
